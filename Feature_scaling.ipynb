Feature Scaling 
Definition: Feature scaling is a data preprocessing technique used to bring numerical features (columns) into a similar range or scale.

In simple words, feature scaling means converting different numerical values into a common scale so that no feature dominates others because of its large values.
Why do we use feature scaling?
- Improves machine learning model performance.
- Prevents large-valued features from dominating.
- Helps algorithms such as KNN, K-Means and SVM.



Min-Max Scaling
Definition: Min-Max Scaling is a feature scaling technique that transforms numerical values into a fixed range, usually between 0 and 1.

Formula:
\[
X_{scaled}=\frac{X-X_{min}}{X_{max}-X_{min}}
\]

Where:
- X = Original value
- Xmin = Minimum value
- Xmax = Maximum value

```python
data={
    "Name":['Akhil','Rahul','John','Anu','David'],
    "Age":[20,22,21,23,24],
    "Study_hrs":[2,4,3,5,6],
    "Marks":[450,780,650,900,850],
    "Salary":[15000,25000,20000,30000,18000]
}
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
#Create minmaxscalar object
Scalar=MinMaxScaler()
#select numerical columns 
Columns=['Age','Study_hrs','Marks','Salary']
#Apply minmax scaling 
Scaled_data=Scalar.fit_transform(df[Columns])
```

```python
Scaled_data
```

```python
#Convert dataframe 
Scaled_df=pd.DataFrame(Scaled_data,columns=Columns)
```

```python
Scaled_df
```

Standardization (Z-Score Scaling)
Definition: Standardization is a feature scaling technique used in data preprocessing to transform numerical data so that it has a mean of 0 and a standard deviation of 1.

In simple words, standardization changes the values based on how far they are from the average (mean).
Why do we use Standardization?
1. To bring numerical features to a common scale.
2. To improve machine learning model performance.
3. To prevent features with large values from dominating.
4. Useful for algorithms like KNN, SVM, Logistic Regression and PCA.

Difference between Min-Max Scaling and Standardization
Min-Max Scaling	Standardization
Range usually 0 to 1	Mean = 0
Uses minimum and maximum	Uses mean and standard deviation
Sensitive to outliers	Also sensitive to outliers
Uses MinMaxScaler	Uses StandardScaler

```python
from sklearn.preprocessing import StandardScaler
#create Standard scalar object 
scalar=StandardScaler()
#Select numerical columns 
Columns=['Age','Study_hrs','Marks','Salary']
#apply standardization
Scaled_data=scalar.fit_transform(df[Columns])
```

```python
Scaled_data
```

```python
#convert result into dataframe 
standard_df=pd.DataFrame(Scaled_data,columns=Columns)
```
